## Учёт заявок в IT-поддержке”

В компании есть небольшая IT-поддержка. Сотрудники отправляют заявки: `не работает принтер`, `забыл пароль`, `нет интернета`. Нужно сделать консольную программу, которая помогает хранить и обрабатывать эти заявки.

### Что должна уметь программа

Меню:
1. Добавить заявку
2. Показать все заявки
3. Изменить статус заявки
4. Показать открытые заявки
5. Найти заявку по имени сотрудника
0. Выход

### Данные одной заявки
```txt
Номер заявки
Имя сотрудника
Описание проблемы
Статус
```

**Статусы можно сделать такими:**
`новая`, `в работе`, `закрыта`


### Пример заявки
```txt
№1
Сотрудник: Иван
Проблема: Не работает принтер
Статус: новая
```

### Обязательные условия
1. Номер заявки создаётся автоматически.
2. При добавлении заявки статус всегда `новая`.
3. Можно изменить статус только на:
    - `новая`
    - `в работе`
    - `закрыта`
4. При поиске по имени программа должна показывать все заявки этого сотрудника.
5. Открытые заявки — это заявки со статусом `новая` или `в работе`.
6. Решение должно быть реализовано в объектно-ориентированном стиле с соблюдением принципов `SOLID`:
   - отдельный класс для заявки;
   - отдельный класс или сервис для управления заявками;
   - отдельный класс для работы с консольным меню;
   - классы не должны брать на себя лишние обязанности;
   - зависимости между частями программы должны быть минимальными и понятными.

````python
from dataclasses import dataclass

@dataclass 
class Application:
    id:int
    name:str
    problem_desc:str
    status:str = 'new'

class Request:
    def __init__(self):
        self.cases: dict[int, Application] = {} 
        self.statuses = {'new','in_progress','closed'}
        self.next_ticket = 0
        

    def add_request(self,name,problem_desc,status):
        current_ti = self.next_ticket + 1

        if self.cases.get(current_ti):
            raise ValueError('already exists')
        
        

        self.cases[current_ti] = Application(
            current_ti,
            name,
            problem_desc,
            status
        )
        self.next_ticket += 1
        

    def show_requsets(self):
        for _, case in self.cases.items():
            print(f'{case.id} | {case.name} | {case.problem_desc} | {case.status}')
    
    def change_status(self,number):
        if self.cases[number].status == 'new':
            self.cases[number].status = 'in_progress'
            return
        
        if self.cases[number].status == 'in_progress':
            self.cases[number].status = 'closed'
            return
            


    def show_opened(self):
        opened = []

        for _,case in self.cases.items():
            if case.status != 'closed':
                opened.append(case)
        print(opened)
    
    def find_by_name(self,name):
        for _,case in self.cases:
            if case.name == name:
                return print(case)
        return None
    
    def find_ticket_by_id(self,id):
        for _,case in self.cases.items():
            if case.id == id:
                return case
        return None


class Menu:
    

    def main(self,request = Request()):
        print('-------------MENU-------------')

        while True:
            user_choise = int(input('choose option: \n 1.create request \n 2.show all requests \n 3.change status \n 4.show opened \n 5.find request by name \n 6.Exit'))

            if user_choise == 1:
                
                name = input('write a name')
                p_description = input('write a description')
                status = 'new'
                request.add_request(name,p_description,status)

            if user_choise == 2:
                request.show_requsets()

            if user_choise == 3:
                ticket_id = int(input('write an id to change'))
                if request.find_ticket_by_id(ticket_id):

                    request.change_status(ticket_id)
            
            if user_choise == 4:
                request.show_opened()

            if user_choise == 5:
                
                request.find_by_name(name)

            if user_choise == 6:
                break

menu = Menu()
menu.main()

            







    
        
````
